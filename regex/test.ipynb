```python
decode_url_pattern = r"""
    (?P<protocol>([Ff][Tt][Pp])|[Hh][Tt][Tt][Pp][Ss]?)://
    ((?P<user>\w+):(?P<password>\w+)@)?
    (?P<hostname>[\w\.]+)?
    (?P<port>:\d+)?
    (?P<path>[\w/\.]+)
    (?P<query>\??[\w=&]+)?
    (?P<fragment>\#\w+)?
"""

# url_string = "https://google.com/search?q=regEx"
# url_string = "ftp://ftp.someserver.com/data/wheather.json"
url_string = "https://www.w3schools.com/python/python_regex.asp#Metacharacters"

print("full string: ", re.findall(decode_url_pattern, url_string, re.VERBOSE))
print("protocol: ", re.match(decode_url_pattern, url_string, re.VERBOSE).group("protocol"))
print("user: ", re.match(decode_url_pattern, url_string, re.VERBOSE).group("user"))
print("password: ", re.match(decode_url_pattern, url_string, re.VERBOSE).group("password"))
print("hostname: ", re.match(decode_url_pattern, url_string, re.VERBOSE).group("hostname"))
print("port: ", re.match(decode_url_pattern, url_string, re.VERBOSE).group("port"))
print("path: ", re.match(decode_url_pattern, url_string, re.VERBOSE).group("path"))
print("query: ", re.match(decode_url_pattern, url_string, re.VERBOSE).group("query"))
print("fragment: ", re.match(decode_url_pattern, url_string, re.VERBOSE).group("fragment"))
```
